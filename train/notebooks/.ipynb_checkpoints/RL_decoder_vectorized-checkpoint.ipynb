```python

# ============================================================
# CELL 1 — IMPORTS AND CONFIGURATION
# Vectorized Reinforcement Learning for Quantum Error Correction
# ============================================================

import random
import numpy as np
import stim
import torch

# ------------------------------------------------------------
# Reproducibility
# ------------------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# ------------------------------------------------------------
# Hardware
# ------------------------------------------------------------

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# ------------------------------------------------------------
# Surface-code development configuration
# ------------------------------------------------------------

DISTANCE = 3
ROUNDS = 3
NOISE_RATE = 0.01

# ------------------------------------------------------------
# Vectorized environment configuration
# ------------------------------------------------------------

# Number of independent episodes processed together.
NUM_ENVS = 256

# A correction can touch all d^2 data qubits in the
# initial development environment, followed by STOP.
NUM_DATA_QUBITS = DISTANCE**2
STOP_ACTION = NUM_DATA_QUBITS
NUM_ACTIONS = NUM_DATA_QUBITS + 1
MAX_STEPS = NUM_DATA_QUBITS + 1

# ------------------------------------------------------------
# Display configuration
# ------------------------------------------------------------

print("=" * 70)
print("VECTORISED RL FOR QUANTUM ERROR CORRECTION")
print("=" * 70)

print(f"Random seed             : {SEED}")
print(f"Surface-code distance   : {DISTANCE}")
print(f"Measurement rounds      : {ROUNDS}")
print(f"Noise rate               : {NOISE_RATE}")
print(f"Number of parallel envs  : {NUM_ENVS}")
print(f"Data-qubit count         : {NUM_DATA_QUBITS}")
print(f"Action count             : {NUM_ACTIONS}")
print(f"STOP action              : {STOP_ACTION}")
print(f"Maximum episode steps    : {MAX_STEPS}")
print(f"PyTorch device           : {DEVICE}")

if torch.cuda.is_available():
    print(f"GPU                      : {torch.cuda.get_device_name(0)}")

print("=" * 70)

assert NUM_ENVS > 0
assert NUM_DATA_QUBITS == DISTANCE**2
assert NUM_ACTIONS == NUM_DATA_QUBITS + 1
assert MAX_STEPS >= NUM_DATA_QUBITS + 1

print("Configuration checks passed.")
```

```python

# ============================================================
# CELL 2 — BUILD THE ROTATED SURFACE-CODE CIRCUIT
# ============================================================

def build_surface_code_circuit(
    distance: int,
    rounds: int,
    noise_rate: float,
) -> stim.Circuit:
    """
    Build a noisy rotated surface-code memory-X circuit.

    Noise is applied through Stim's generated-circuit options.
    """
    return stim.Circuit.generated(
        "surface_code:rotated_memory_x",
        distance=distance,
        rounds=rounds,
        after_clifford_depolarization=noise_rate,
        before_round_data_depolarization=noise_rate,
        before_measure_flip_probability=noise_rate,
        after_reset_flip_probability=noise_rate,
    )


circuit = build_surface_code_circuit(
    distance=DISTANCE,
    rounds=ROUNDS,
    noise_rate=NOISE_RATE,
)

# ------------------------------------------------------------
# Inspect circuit dimensions
# ------------------------------------------------------------

NUM_QUBITS = circuit.num_qubits
NUM_DETECTORS = circuit.num_detectors
NUM_OBSERVABLES = circuit.num_observables
NUM_MEASUREMENTS = circuit.num_measurements

print("=" * 70)
print("SURFACE-CODE CIRCUIT SUMMARY")
print("=" * 70)

print(f"Code family          : Rotated surface code")
print(f"Memory basis         : X")
print(f"Distance             : {DISTANCE}")
print(f"Measurement rounds   : {ROUNDS}")
print(f"Noise rate           : {NOISE_RATE}")
print(f"Qubit slots          : {NUM_QUBITS}")
print(f"Detector count       : {NUM_DETECTORS}")
print(f"Observable count     : {NUM_OBSERVABLES}")
print(f"Measurement count    : {NUM_MEASUREMENTS}")

print("\nFirst part of the generated circuit:")
print(str(circuit)[:1500])

# ------------------------------------------------------------
# Basic consistency checks
# ------------------------------------------------------------

assert NUM_DETECTORS > 0
assert NUM_OBSERVABLES >= 1
assert NUM_MEASUREMENTS > 0

print("\nCircuit construction checks passed.")
```

```python
# ============================================================
# CELL 3 — CLEAN REFERENCE CIRCUIT AND DATA-QUBIT IDENTIFICATION
# ============================================================

clean_circuit = build_surface_code_circuit(
    distance=DISTANCE,
    rounds=ROUNDS,
    noise_rate=0.0,
)

# In the generated rotated memory-X circuit, the initial RX
# instruction resets the data qubits into the X basis.
data_qubits = None

for instruction in clean_circuit:
    if isinstance(instruction, stim.CircuitInstruction):
        if instruction.name == "RX":
            data_qubits = [
                target.value
                for target in instruction.targets_copy()
            ]
            break

assert data_qubits is not None, "Could not identify data qubits."
assert len(data_qubits) == NUM_DATA_QUBITS, (
    f"Expected {NUM_DATA_QUBITS} data qubits, "
    f"but found {len(data_qubits)}."
)
assert len(set(data_qubits)) == NUM_DATA_QUBITS

print("=" * 70)
print("CLEAN REFERENCE CIRCUIT")
print("=" * 70)
print(f"Distance                : {DISTANCE}")
print(f"Measurement rounds      : {ROUNDS}")
print(f"Qubit slots              : {clean_circuit.num_qubits}")
print(f"Detector count           : {clean_circuit.num_detectors}")
print(f"Observable count         : {clean_circuit.num_observables}")
print(f"Measurement count        : {clean_circuit.num_measurements}")
print(f"Number of data qubits    : {len(data_qubits)}")
print(f"Data-qubit Stim IDs      : {data_qubits}")

# Check that the clean circuit has no nonzero-probability noise.
noise_instructions = []

for instruction in clean_circuit.flattened():
    if isinstance(instruction, stim.CircuitInstruction):
        if instruction.gate_args_copy():
            noise_instructions.append(
                (
                    instruction.name,
                    instruction.gate_args_copy()
                )
            )

print(f"\nInstructions with gate arguments: {len(noise_instructions)}")
print("\nFirst part of the clean circuit:")
print(str(clean_circuit)[:1000])

assert clean_circuit.num_detectors == NUM_DETECTORS
assert clean_circuit.num_observables == NUM_OBSERVABLES
assert clean_circuit.num_measurements == NUM_MEASUREMENTS

print("\nClean-circuit checks passed.")
```

```python
# ============================================================
# CELL 4 — BUILD THE CONTROLLED ERROR DATASET
# ============================================================

def inject_deterministic_z_error(
    base_circuit: stim.Circuit,
    qubit_ids: list[int],
) -> stim.Circuit:
    """
    Insert deterministic Z errors immediately after the initial
    RX operation, before the stabilizer-measurement rounds.
    """
    lines = str(base_circuit).splitlines()

    rx_line_index = next(
        (
            i for i, line in enumerate(lines)
            if line.strip().startswith("RX ")
        ),
        None,
    )

    if rx_line_index is None:
        raise ValueError("Could not find the initial RX instruction.")

    if len(qubit_ids) == 0:
        return base_circuit.copy()

    error_line = (
        "Z_ERROR(1) "
        + " ".join(str(q) for q in qubit_ids)
    )

    lines.insert(rx_line_index + 1, error_line)

    return stim.Circuit("\n".join(lines))


# ------------------------------------------------------------
# 1. Obtain the clean reference detector/observable record.
# ------------------------------------------------------------

reference_sampler = clean_circuit.compile_detector_sampler()

reference_detectors, reference_observables = reference_sampler.sample(
    shots=1,
    separate_observables=True,
)

reference_detectors = reference_detectors[0].astype(np.uint8)
reference_observables = reference_observables[0].astype(np.uint8)

print("Reference detector record:", reference_detectors.shape)
print("Reference observable record:", reference_observables.shape)
print("Reference detector weight:", int(reference_detectors.sum()))
print("Reference observable values:", reference_observables)


# ------------------------------------------------------------
# 2. Measure the response to each individual data-qubit Z error.
# ------------------------------------------------------------

single_detector_effects = np.zeros(
    (NUM_DATA_QUBITS, NUM_DETECTORS),
    dtype=np.uint8,
)

single_observable_effects = np.zeros(
    (NUM_DATA_QUBITS, NUM_OBSERVABLES),
    dtype=np.uint8,
)

for action, qubit_id in enumerate(data_qubits):
    fault_circuit = inject_deterministic_z_error(
        clean_circuit,
        [qubit_id],
    )

    fault_sampler = fault_circuit.compile_detector_sampler()

    fault_detectors, fault_observables = fault_sampler.sample(
        shots=1,
        separate_observables=True,
    )

    # Compare with the clean reference record.
    single_detector_effects[action] = (
        fault_detectors[0].astype(np.uint8)
        ^ reference_detectors
    )

    single_observable_effects[action] = (
        fault_observables[0].astype(np.uint8)
        ^ reference_observables
    )


# ------------------------------------------------------------
# 3. Enumerate all 2^9 possible physical error patterns.
# ------------------------------------------------------------

NUM_ERROR_PATTERNS = 2**NUM_DATA_QUBITS

error_patterns = (
    (
        np.arange(NUM_ERROR_PATTERNS, dtype=np.uint16)[:, None]
        >> np.arange(NUM_DATA_QUBITS, dtype=np.uint16)[None, :]
    )
    & 1
).astype(np.uint8)

# The detector response is the XOR of the responses of the
# individual Z errors in the pattern.
detector_syndromes = (
    error_patterns @ single_detector_effects
) % 2

logical_observables = (
    error_patterns @ single_observable_effects
) % 2

detector_syndromes = detector_syndromes.astype(np.uint8)
logical_observables = logical_observables.astype(np.uint8)

physical_weights = error_patterns.sum(axis=1).astype(np.uint8)
syndrome_weights = detector_syndromes.sum(axis=1).astype(np.uint8)

# Keep the dataset in a dictionary for convenient access.
controlled_dataset = {
    "error_patterns": error_patterns,
    "physical_weights": physical_weights,
    "detectors": detector_syndromes,
    "syndrome_weights": syndrome_weights,
    "logical_observables": logical_observables,
}

print("\n" + "=" * 70)
print("CONTROLLED DATASET SUMMARY")
print("=" * 70)
print(f"Number of error patterns       : {len(error_patterns)}")
print(f"Error-pattern shape            : {error_patterns.shape}")
print(f"Detector-syndrome shape        : {detector_syndromes.shape}")
print(f"Logical-observable shape       : {logical_observables.shape}")
print(f"Unique detector syndromes      : {len(np.unique(detector_syndromes, axis=0))}")
print(f"Patterns with no errors        : {int(np.sum(physical_weights == 0))}")
print(f"Patterns with nonzero syndrome : {int(np.sum(syndrome_weights > 0))}")
print(f"Patterns with logical flip     : {int(np.sum(logical_observables[:, 0] == 1))}")

print("\nFirst five patterns:")
for i in range(5):
    print(
        f"Pattern {i:3d}: "
        f"errors={error_patterns[i].tolist()}, "
        f"weight={int(physical_weights[i])}, "
        f"syndrome_weight={int(syndrome_weights[i])}, "
        f"logical={logical_observables[i].tolist()}"
    )

# ------------------------------------------------------------
# 4. Basic validation.
# ------------------------------------------------------------

assert error_patterns.shape == (NUM_ERROR_PATTERNS, NUM_DATA_QUBITS)
assert detector_syndromes.shape == (NUM_ERROR_PATTERNS, NUM_DETECTORS)
assert logical_observables.shape == (NUM_ERROR_PATTERNS, NUM_OBSERVABLES)

# Pattern zero must have no effect relative to the clean reference.
assert np.array_equal(error_patterns[0], np.zeros(NUM_DATA_QUBITS, dtype=np.uint8))
assert np.array_equal(detector_syndromes[0], np.zeros(NUM_DETECTORS, dtype=np.uint8))
assert np.array_equal(logical_observables[0], np.zeros(NUM_OBSERVABLES, dtype=np.uint8))

# Each physical error pattern must have a unique binary representation.
assert len(np.unique(error_patterns, axis=0)) == NUM_ERROR_PATTERNS

print("\nControlled dataset checks passed.")
```

```python
# ============================================================
# CELL 5 — ANALYSE SYNDROME AMBIGUITY
# ============================================================

syndrome_classes = {}

for i in range(NUM_ERROR_PATTERNS):
    syndrome_key = tuple(detector_syndromes[i].tolist())

    if syndrome_key not in syndrome_classes:
        syndrome_classes[syndrome_key] = {
            "count": 0,
            "logical_values": set(),
            "minimum_weight": NUM_DATA_QUBITS + 1,
            "maximum_weight": -1,
        }

    entry = syndrome_classes[syndrome_key]
    entry["count"] += 1
    entry["logical_values"].add(int(logical_observables[i, 0]))
    entry["minimum_weight"] = min(
        entry["minimum_weight"],
        int(physical_weights[i]),
    )
    entry["maximum_weight"] = max(
        entry["maximum_weight"],
        int(physical_weights[i]),
    )

ambiguous_classes = {
    syndrome: info
    for syndrome, info in syndrome_classes.items()
    if len(info["logical_values"]) > 1
}

unambiguous_classes = {
    syndrome: info
    for syndrome, info in syndrome_classes.items()
    if len(info["logical_values"]) == 1
}

print("=" * 70)
print("SYNDROME AMBIGUITY ANALYSIS")
print("=" * 70)
print(f"Unique syndrome classes       : {len(syndrome_classes)}")
print(f"Ambiguous syndrome classes    : {len(ambiguous_classes)}")
print(f"Unambiguous syndrome classes  : {len(unambiguous_classes)}")

if ambiguous_classes:
    print("\nExample ambiguous syndrome classes:")
    for syndrome, info in list(ambiguous_classes.items())[:5]:
        print(
            f"  Syndrome weight={sum(syndrome):2d}, "
            f"patterns={info['count']:3d}, "
            f"logical outcomes={sorted(info['logical_values'])}, "
            f"error weights={info['minimum_weight']}-"
            f"{info['maximum_weight']}"
        )

# A useful check: the zero syndrome should include the no-error pattern.
zero_syndrome = tuple([0] * NUM_DETECTORS)
assert zero_syndrome in syndrome_classes

print("\nZero-syndrome class:")
print(syndrome_classes[zero_syndrome])

print("\nAmbiguity analysis complete.")
```

```python
'''

Why this matters for the RL environment
We need to distinguish three things:
1. Observation: the detector syndrome, which the agent can see.
2. Correction frame: the nine-bit vector of corrections the agent has applied.
3. Hidden error: the actual nine-bit error pattern, used internally to calculate the reward.

'''
```

```python

# ============================================================
# CELL 6 — VECTORISED RL ENVIRONMENT
# ============================================================

class VectorizedQECEnv:
    """
    Batched environment for controlled Z-error correction.

    Observation:
        [detector syndrome, current correction frame]

    Actions:
        0 ... NUM_DATA_QUBITS-1 : toggle a Z correction
        STOP_ACTION              : stop and evaluate correction

    Rewards:
        +1 : successful correction
        -1 : unsuccessful correction
         0 : nonterminal step

    The hidden physical error is never included in observations.
    """

    def __init__(
        self,
        dataset,
        single_detector_effects,
        single_observable_effects,
        num_envs,
        max_steps,
        stop_action,
        seed=42,
    ):
        self.dataset = dataset
        self.single_detector_effects = np.asarray(
            single_detector_effects, dtype=np.uint8
        )
        self.single_observable_effects = np.asarray(
            single_observable_effects, dtype=np.uint8
        )

        self.num_envs = num_envs
        self.max_steps = max_steps
        self.stop_action = stop_action
        self.num_actions = stop_action + 1

        self.num_data_qubits = stop_action
        self.num_detectors = self.single_detector_effects.shape[1]
        self.num_observables = self.single_observable_effects.shape[1]
        self.observation_dim = self.num_detectors + self.num_data_qubits

        self.rng = np.random.default_rng(seed)

        # Hidden physical errors and their measured effects.
        self.error_indices = np.zeros(num_envs, dtype=np.int64)
        self.hidden_errors = np.zeros(
            (num_envs, self.num_data_qubits), dtype=np.uint8
        )
        self.hidden_logical = np.zeros(
            (num_envs, self.num_observables), dtype=np.uint8
        )

        # Agent-controlled correction frames.
        self.corrections = np.zeros_like(self.hidden_errors)

        # Per-environment episode bookkeeping.
        self.steps = np.zeros(num_envs, dtype=np.int32)
        self.done = np.ones(num_envs, dtype=bool)
        self.episode_success = np.zeros(num_envs, dtype=bool)

        self.detectors = np.zeros(
            (num_envs, self.num_detectors), dtype=np.uint8
        )

    def _get_observation(self):
        """Return only the agent-visible information."""
        return np.concatenate(
            [self.detectors, self.corrections],
            axis=1,
        ).astype(np.float32)

    def reset(self):
        """Start a fresh episode in every parallel environment."""
        n_patterns = len(self.dataset["error_patterns"])

        self.error_indices = self.rng.integers(
            0, n_patterns, size=self.num_envs
        )

        # These arrays are internal state, not part of the observation.
        self.hidden_errors = self.dataset["error_patterns"][
            self.error_indices
        ].copy()

        self.hidden_logical = self.dataset["logical_observables"][
            self.error_indices
        ].copy()

        self.detectors = self.dataset["detectors"][
            self.error_indices
        ].copy()

        self.corrections.fill(0)
        self.steps.fill(0)
        self.done.fill(False)
        self.episode_success.fill(False)

        return self._get_observation()

    def _evaluate_success(self, env_indices):
        """Evaluate terminal episodes using hidden environment state."""
        if len(env_indices) == 0:
            return

        correction_syndromes = (
            self.corrections[env_indices]
            @ self.single_detector_effects
        ) % 2

        correction_logical = (
            self.corrections[env_indices]
            @ self.single_observable_effects
        ) % 2

        residual_syndromes = (
            self.detectors[env_indices] ^ correction_syndromes
        )

        residual_logical = (
            self.hidden_logical[env_indices] ^ correction_logical
        )

        syndrome_ok = np.all(residual_syndromes == 0, axis=1)
        logical_ok = np.all(residual_logical == 0, axis=1)

        self.episode_success[env_indices] = syndrome_ok & logical_ok

    def step(self, actions):
        """
        Advance every active environment by one step.

        Parameters
        ----------
        actions : array of shape (num_envs,)
            One discrete action per environment.

        Returns
        -------
        observation, rewards, done, info
        """
        actions = np.asarray(actions, dtype=np.int64)

        if actions.shape != (self.num_envs,):
            raise ValueError(
                f"Expected actions of shape ({self.num_envs},), "
                f"got {actions.shape}."
            )

        if np.any((actions < 0) | (actions >= self.num_actions)):
            raise ValueError("Action index out of range.")

        rewards = np.zeros(self.num_envs, dtype=np.float32)

        active = ~self.done
        active_indices = np.flatnonzero(active)

        if len(active_indices) == 0:
            return (
                self._get_observation(),
                rewards,
                self.done.copy(),
                {"success": self.episode_success.copy()},
            )

        active_actions = actions[active_indices]

        # Increment the step counter only for active environments.
        self.steps[active_indices] += 1

        # Apply correction toggles to environments that did not STOP.
        toggle_mask = active_actions != self.stop_action
        toggle_envs = active_indices[toggle_mask]
        toggle_actions = active_actions[toggle_mask]

        if len(toggle_envs) > 0:
            self.corrections[toggle_envs, toggle_actions] ^= 1

        # STOP terminates an episode.
        stop_envs = active_indices[
            active_actions == self.stop_action
        ]

        if len(stop_envs) > 0:
            self._evaluate_success(stop_envs)
            self.done[stop_envs] = True
            rewards[stop_envs] = np.where(
                self.episode_success[stop_envs], 1.0, -1.0
            )

        # Enforce the maximum episode length.
        timed_out = active_indices[
            (self.steps[active_indices] >= self.max_steps)
            & (~self.done[active_indices])
        ]

        if len(timed_out) > 0:
            self._evaluate_success(timed_out)
            self.done[timed_out] = True
            rewards[timed_out] = np.where(
                self.episode_success[timed_out], 1.0, -1.0
            )

        info = {
            "success": self.episode_success.copy(),
        }

        return (
            self._get_observation(),
            rewards,
            self.done.copy(),
            info,
        )


# ------------------------------------------------------------
# Instantiate the vectorized environment.
# ------------------------------------------------------------

env = VectorizedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=NUM_ENVS,
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=SEED,
)

initial_observation = env.reset()

print("=" * 70)
print("VECTORIZED ENVIRONMENT INITIALIZED")
print("=" * 70)
print(f"Parallel environments : {env.num_envs}")
print(f"Observation dimension : {env.observation_dim}")
print(f"Action count          : {env.num_actions}")
print(f"STOP action           : {env.stop_action}")
print(f"Observation shape     : {initial_observation.shape}")
print(f"Correction shape      : {env.corrections.shape}")
print(f"All episodes active   : {np.all(~env.done)}")
print(f"Observation dtype     : {initial_observation.dtype}")

assert initial_observation.shape == (
    NUM_ENVS, NUM_DETECTORS + NUM_DATA_QUBITS
)
assert np.all(initial_observation[:, NUM_DETECTORS:] == 0)
assert np.all(~env.done)

print("\nVectorized environment initialization passed.")
```

```python
'''

The agent's observation is
\[
o_t=[D,C_t],
\]

where \(D\) is the observed detector syndrome and \(C_t\) is the correction frame. The hidden physical error \(E\) is stored internally and never returned in the observation.

At termination, the decoder succeeds only when both conditions hold:
\[
D(E)\oplus D(C)=0
\]

and
\[
L(E)\oplus L(C)=0.
\]

Here, \(D(\cdot)\) denotes the detector response and \(L(\cdot)\) the logical-observable effect. The first condition checks syndrome consistency; the second checks logical correctness.

'''
```

```python
'''

The reward must reflect logical correction success, not merely whether the agent has taken an action or matched the syndrome.
We’ll test that with known cases, including a logical failure that has zero syndrome.
No error + STOP: should receive +1.
Zero-syndrome logical error + STOP: should receive -1.
Single-qubit error + matching correction + STOP: should receive +1.
'''




# ============================================================
# CELL 7 — VALIDATE VECTORISED REWARD SEMANTICS
# ============================================================

test_env = VectorizedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=3,
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=SEED,
)

test_env.reset()

# Find a nontrivial error pattern with zero syndrome but logical flip.
logical_failure_candidates = np.flatnonzero(
    (syndrome_weights == 0)
    & (logical_observables[:, 0] == 1)
)

assert len(logical_failure_candidates) > 0
logical_failure_index = int(logical_failure_candidates[0])

# Select one single-qubit error pattern: bit 0 is flipped.
single_error_index = 1
assert error_patterns[single_error_index, 0] == 1
assert physical_weights[single_error_index] == 1

# Environment 0: no physical error.
# Environment 1: zero-syndrome error with a logical flip.
# Environment 2: single-qubit error on data-qubit action 0.
test_indices = np.array(
    [0, logical_failure_index, single_error_index],
    dtype=np.int64,
)

test_env.error_indices = test_indices.copy()
test_env.hidden_errors = error_patterns[test_indices].copy()
test_env.hidden_logical = logical_observables[test_indices].copy()
test_env.detectors = detector_syndromes[test_indices].copy()
test_env.corrections.fill(0)
test_env.steps.fill(0)
test_env.done.fill(False)
test_env.episode_success.fill(False)

print("Chosen test pattern indices:", test_indices)
print(
    "Initial syndrome weights:",
    test_env.detectors.sum(axis=1).tolist(),
)
print(
    "Hidden logical outcomes (test-only):",
    test_env.hidden_logical[:, 0].tolist(),
)

# Step 1:
# Environments 0 and 1 STOP immediately.
# Environment 2 applies correction action 0.
actions = np.array([STOP_ACTION, STOP_ACTION, 0])

obs, rewards, done, info = test_env.step(actions)

print("\nAfter first step:")
print("Rewards:", rewards.tolist())
print("Done:", done.tolist())
print("Correction frames:", test_env.corrections.tolist())

assert rewards[0] == 1.0
assert rewards[1] == -1.0
assert rewards[2] == 0.0
assert done.tolist() == [True, True, False]
assert test_env.corrections[2, 0] == 1

# Step 2: environment 2 now stops after applying the matching correction.
obs, rewards, done, info = test_env.step(
    np.array([STOP_ACTION, STOP_ACTION, STOP_ACTION])
)

print("\nAfter second step:")
print("Rewards:", rewards.tolist())
print("Done:", done.tolist())
print("Success flags:", info["success"].tolist())

assert rewards[2] == 1.0
assert done.tolist() == [True, True, True]
assert info["success"].tolist() == [True, False, True]

print("\nAll reward-semantic tests passed.")
```

```python

# ============================================================
# CELL 8 — BASELINE EVALUATION
# ============================================================

# ------------------------------------------------------------
# 1. Exact no-correction baseline over all 512 patterns.
# ------------------------------------------------------------

baseline_env = VectorizedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=NUM_ERROR_PATTERNS,
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=SEED,
)

# Load every possible error pattern exactly once.
baseline_env.error_indices = np.arange(NUM_ERROR_PATTERNS)
baseline_env.hidden_errors = error_patterns.copy()
baseline_env.hidden_logical = logical_observables.copy()
baseline_env.detectors = detector_syndromes.copy()
baseline_env.corrections.fill(0)
baseline_env.steps.fill(0)
baseline_env.done.fill(False)
baseline_env.episode_success.fill(False)

_, baseline_rewards, baseline_done, baseline_info = baseline_env.step(
    np.full(NUM_ERROR_PATTERNS, STOP_ACTION)
)

no_correction_successes = int(baseline_info["success"].sum())
no_correction_rate = no_correction_successes / NUM_ERROR_PATTERNS

assert np.all(baseline_done)

print("=" * 70)
print("NO-CORRECTION BASELINE")
print("=" * 70)
print(f"Patterns evaluated : {NUM_ERROR_PATTERNS}")
print(f"Successful         : {no_correction_successes}")
print(f"Success rate       : {100 * no_correction_rate:.2f}%")


# ------------------------------------------------------------
# 2. Random policy over multiple vectorized batches.
# ------------------------------------------------------------

NUM_RANDOM_BATCHES = 20
RANDOM_EPISODES = NUM_RANDOM_BATCHES * NUM_ENVS

random_env = VectorizedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=NUM_ENVS,
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=SEED + 1,
)

policy_rng = np.random.default_rng(SEED + 2)

random_successes = 0
random_rewards_sum = 0.0
random_episode_lengths = []

for batch in range(NUM_RANDOM_BATCHES):
    random_env.reset()

    batch_lengths = np.zeros(NUM_ENVS, dtype=np.int32)

    for _ in range(MAX_STEPS):
        actions = policy_rng.integers(
            low=0,
            high=NUM_ACTIONS,
            size=NUM_ENVS,
        )

        _, rewards, done, info = random_env.step(actions)
        batch_lengths[batch_lengths < MAX_STEPS] += (
            (~done) | (rewards != 0)
        ).astype(np.int32)

        random_rewards_sum += float(rewards.sum())

        if np.all(done):
            break

    random_successes += int(random_env.episode_success.sum())
    random_episode_lengths.extend(batch_lengths.tolist())

random_success_rate = random_successes / RANDOM_EPISODES
random_mean_reward = random_rewards_sum / RANDOM_EPISODES

print("\n" + "=" * 70)
print("RANDOM-POLICY BASELINE")
print("=" * 70)
print(f"Episodes evaluated : {RANDOM_EPISODES}")
print(f"Successful         : {random_successes}")
print(f"Success rate       : {100 * random_success_rate:.2f}%")
print(f"Mean terminal reward: {random_mean_reward:.4f}")

print("\nBaseline evaluation complete.")
```

```python
# ============================================================
# CELL 9 — MINIMUM-WEIGHT SYNDROME LOOKUP BASELINE
# ============================================================

# Map each syndrome to a minimum-weight correction pattern.
correction_lookup = {}

for i in range(NUM_ERROR_PATTERNS):
    syndrome_key = tuple(detector_syndromes[i].tolist())
    weight = int(physical_weights[i])

    if (
        syndrome_key not in correction_lookup
        or weight < correction_lookup[syndrome_key]["weight"]
    ):
        correction_lookup[syndrome_key] = {
            "correction": error_patterns[i].copy(),
            "weight": weight,
            "logical": logical_observables[i].copy(),
        }

# Evaluate on all 512 physical error patterns.
lookup_success = np.zeros(NUM_ERROR_PATTERNS, dtype=bool)

for i in range(NUM_ERROR_PATTERNS):
    syndrome_key = tuple(detector_syndromes[i].tolist())
    correction = correction_lookup[syndrome_key]["correction"]

    correction_syndrome = (
        correction @ single_detector_effects
    ) % 2

    correction_logical = (
        correction @ single_observable_effects
    ) % 2

    residual_syndrome = (
        detector_syndromes[i] ^ correction_syndrome
    )

    residual_logical = (
        logical_observables[i] ^ correction_logical
    )

    lookup_success[i] = (
        np.all(residual_syndrome == 0)
        and np.all(residual_logical == 0)
    )

lookup_success_count = int(lookup_success.sum())
lookup_success_rate = lookup_success_count / NUM_ERROR_PATTERNS

print("=" * 70)
print("MINIMUM-WEIGHT LOOKUP BASELINE")
print("=" * 70)
print(f"Patterns evaluated        : {NUM_ERROR_PATTERNS}")
print(f"Unique syndromes covered  : {len(correction_lookup)}")
print(f"Successful corrections   : {lookup_success_count}")
print(f"Success rate              : {100 * lookup_success_rate:.2f}%")

print("\nComparison:")
print(f"No correction             : {100 * no_correction_rate:.2f}%")
print(f"Random policy             : {100 * random_success_rate:.2f}%")
print(f"Minimum-weight lookup     : {100 * lookup_success_rate:.2f}%")

assert len(correction_lookup) == len(syndrome_classes)
print("\nLookup baseline evaluation complete.")
```

```python

# ============================================================
# CELL 11 — DQN AGENT, REPLAY BUFFER, AND TARGET NETWORK
# ============================================================

import random
from collections import deque

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim

# ------------------------------------------------------------
# 1. Reproducibility and device selection
# ------------------------------------------------------------

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
    DEVICE = torch.device("cuda")
else:
    DEVICE = torch.device("cpu")

OBS_DIM = controlled_dataset["detectors"].shape[1] + NUM_DATA_QUBITS
ACTION_DIM = NUM_ACTIONS

print("=" * 70)
print("DQN INITIALIZATION")
print("=" * 70)
print(f"Device       : {DEVICE}")
print(f"Observation  : {OBS_DIM}")
print(f"Actions      : {ACTION_DIM}")
print(f"Hidden error : Not included in observation")


# ------------------------------------------------------------
# 2. Q-network
# ------------------------------------------------------------

class QNetwork(nn.Module):
    """
    Estimates Q(s, a) for every available action.

    Input:
        State vector containing detector bits and correction frame.

    Output:
        One Q-value for each action, including STOP.
    """

    def __init__(self, obs_dim, action_dim, hidden_dim=128):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(obs_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, action_dim),
        )

    def forward(self, states):
        return self.network(states)


# ------------------------------------------------------------
# 3. Experience replay buffer
# ------------------------------------------------------------

class ReplayBuffer:
    """
    Stores transitions:
        state, action, reward, next_state, done
    """

    def __init__(self, capacity, obs_dim, seed=0):
        self.capacity = capacity
        self.rng = np.random.default_rng(seed)

        self.states = np.zeros(
            (capacity, obs_dim), dtype=np.float32
        )
        self.actions = np.zeros(capacity, dtype=np.int64)
        self.rewards = np.zeros(capacity, dtype=np.float32)
        self.next_states = np.zeros(
            (capacity, obs_dim), dtype=np.float32
        )
        self.dones = np.zeros(capacity, dtype=np.float32)

        self.position = 0
        self.size = 0

    def add_batch(
        self, states, actions, rewards, next_states, dones
    ):
        """Store transitions from a vectorized environment."""

        states = np.asarray(states, dtype=np.float32)
        actions = np.asarray(actions, dtype=np.int64)
        rewards = np.asarray(rewards, dtype=np.float32)
        next_states = np.asarray(next_states, dtype=np.float32)
        dones = np.asarray(dones, dtype=np.float32)

        batch_size = len(actions)

        if batch_size > self.capacity:
            raise ValueError("Batch exceeds replay-buffer capacity.")

        indices = (
            self.position + np.arange(batch_size)
        ) % self.capacity

        self.states[indices] = states
        self.actions[indices] = actions
        self.rewards[indices] = rewards
        self.next_states[indices] = next_states
        self.dones[indices] = dones

        self.position = (
            self.position + batch_size
        ) % self.capacity

        self.size = min(self.size + batch_size, self.capacity)

    def sample(self, batch_size):
        """Return a randomly sampled minibatch as NumPy arrays."""

        if self.size < batch_size:
            raise ValueError("Not enough transitions to sample.")

        indices = self.rng.choice(
            self.size, size=batch_size, replace=False
        )

        return (
            self.states[indices],
            self.actions[indices],
            self.rewards[indices],
            self.next_states[indices],
            self.dones[indices],
        )

    def __len__(self):
        return self.size


# ------------------------------------------------------------
# 4. DQN agent
# ------------------------------------------------------------

class DQNAgent:
    def __init__(
        self,
        obs_dim,
        action_dim,
        learning_rate=1e-3,
        gamma=0.99,
        hidden_dim=128,
        seed=0,
    ):
        self.obs_dim = obs_dim
        self.action_dim = action_dim
        self.gamma = gamma
        self.rng = np.random.default_rng(seed)

        # Online network: updated by gradient descent.
        self.online_net = QNetwork(
            obs_dim, action_dim, hidden_dim
        ).to(DEVICE)

        # Target network: provides more stable learning targets.
        self.target_net = QNetwork(
            obs_dim, action_dim, hidden_dim
        ).to(DEVICE)

        self.target_net.load_state_dict(
            self.online_net.state_dict()
        )
        self.target_net.eval()

        self.optimizer = optim.Adam(
            self.online_net.parameters(),
            lr=learning_rate,
        )

        self.loss_function = nn.SmoothL1Loss()

    @torch.no_grad()
    def select_actions(self, observations, epsilon=0.0):
        """
        Epsilon-greedy action selection for a batch of environments.

        With probability epsilon, select a random action.
        Otherwise, select the action with the largest predicted Q-value.
        """

        observations = np.asarray(
            observations, dtype=np.float32
        )

        states = torch.as_tensor(
            observations, dtype=torch.float32, device=DEVICE
        )

        q_values = self.online_net(states)
        actions = q_values.argmax(dim=1).cpu().numpy()

        explore = self.rng.random(len(observations)) < epsilon

        num_explore = int(explore.sum())
        if num_explore:
            actions[explore] = self.rng.integers(
                0, self.action_dim, size=num_explore
            )

        return actions.astype(np.int64)

    def train_step(self, replay_buffer, batch_size=128):
        """
        Perform one DQN optimization step.

        Returns the scalar loss, or None if the buffer is too small.
        """

        if len(replay_buffer) < batch_size:
            return None

        (
            states,
            actions,
            rewards,
            next_states,
            dones,
        ) = replay_buffer.sample(batch_size)

        states = torch.as_tensor(
            states, dtype=torch.float32, device=DEVICE
        )
        actions = torch.as_tensor(
            actions, dtype=torch.long, device=DEVICE
        )
        rewards = torch.as_tensor(
            rewards, dtype=torch.float32, device=DEVICE
        )
        next_states = torch.as_tensor(
            next_states, dtype=torch.float32, device=DEVICE
        )
        dones = torch.as_tensor(
            dones, dtype=torch.float32, device=DEVICE
        )

        # Q-value for the action actually taken.
        current_q = self.online_net(states).gather(
            1, actions.unsqueeze(1)
        ).squeeze(1)

        # Bellman target:
        # reward + gamma * max_a Q_target(next_state, a)
        with torch.no_grad():
            next_q = self.target_net(next_states).max(dim=1).values

            target_q = rewards + (
                self.gamma * (1.0 - dones) * next_q
            )

        loss = self.loss_function(current_q, target_q)

        self.optimizer.zero_grad()
        loss.backward()

        # Reduce the risk of excessively large gradients.
        nn.utils.clip_grad_norm_(
            self.online_net.parameters(), max_norm=10.0
        )

        self.optimizer.step()

        return float(loss.item())

    def update_target_network(self):
        """Copy online-network weights into the target network."""

        self.target_net.load_state_dict(
            self.online_net.state_dict()
        )


# ------------------------------------------------------------
# 5. Instantiate components
# ------------------------------------------------------------

REPLAY_CAPACITY = 50_000
BATCH_SIZE = 128
LEARNING_RATE = 1e-3
GAMMA = 0.99
HIDDEN_DIM = 128

replay_buffer = ReplayBuffer(
    capacity=REPLAY_CAPACITY,
    obs_dim=OBS_DIM,
    seed=SEED,
)

agent = DQNAgent(
    obs_dim=OBS_DIM,
    action_dim=ACTION_DIM,
    learning_rate=LEARNING_RATE,
    gamma=GAMMA,
    hidden_dim=HIDDEN_DIM,
    seed=SEED,
)

# ------------------------------------------------------------
# 6. Smoke tests: shapes, actions, replay, and one update
# ------------------------------------------------------------

test_observations = np.zeros(
    (4, OBS_DIM), dtype=np.float32
)

test_actions = agent.select_actions(
    test_observations, epsilon=0.0
)

assert test_actions.shape == (4,)
assert np.all(test_actions >= 0)
assert np.all(test_actions < ACTION_DIM)

# Store a small batch of synthetic transitions to test training.
synthetic_next_states = np.zeros_like(test_observations)
synthetic_rewards = np.zeros(4, dtype=np.float32)
synthetic_dones = np.ones(4, dtype=np.float32)

replay_buffer.add_batch(
    test_observations,
    test_actions,
    synthetic_rewards,
    synthetic_next_states,
    synthetic_dones,
)

assert len(replay_buffer) == 4
assert replay_buffer.sample(2)[0].shape == (2, OBS_DIM)

# A real optimizer update requires at least BATCH_SIZE transitions.
# Add repeated synthetic transitions solely for this implementation test.
replay_buffer.add_batch(
    np.repeat(test_observations, 32, axis=0),
    np.tile(test_actions, 32),
    np.zeros(128, dtype=np.float32),
    np.repeat(synthetic_next_states, 32, axis=0),
    np.ones(128, dtype=np.float32),
)

test_loss = agent.train_step(
    replay_buffer, batch_size=BATCH_SIZE
)

assert test_loss is not None
assert np.isfinite(test_loss)

agent.update_target_network()

print("-" * 70)
print("Q-network output shape       :", (4, ACTION_DIM))
print("Action selection             : passed")
print("Replay-buffer sampling       : passed")
print(f"Test optimization loss       : {test_loss:.6f}")
print("Target-network update        : passed")
print("DQN initialization tests passed.")
print()
print("Note: synthetic transitions were used only for smoke testing.")
print("Clear the replay buffer before actual training.")
```

```python

# ============================================================
# CELL 12 — IMPROVED DQN TRAINING RUN
# ============================================================

# ------------------------------------------------------------
# 1. Fresh replay buffer, agent, and training environment
# ------------------------------------------------------------

replay_buffer = ReplayBuffer(
    capacity=REPLAY_CAPACITY,
    obs_dim=OBS_DIM,
    seed=SEED + 22,
)

agent = DQNAgent(
    obs_dim=OBS_DIM,
    action_dim=ACTION_DIM,
    learning_rate=LEARNING_RATE,
    gamma=GAMMA,
    hidden_dim=HIDDEN_DIM,
    seed=SEED + 22,
)

env = VectorizedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=NUM_ENVS,
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=SEED + 22,
)

# ------------------------------------------------------------
# 2. Training configuration
# ------------------------------------------------------------

TRAIN_BATCHES = 40
EVAL_EPISODES = 2048
LEARNING_STARTS = 512
GRADIENT_UPDATES_PER_STEP = 4
TARGET_UPDATE_EVERY = 500

EPSILON_START = 1.0
EPSILON_END = 0.05
EPSILON_DECAY_STEPS = 100_000

global_step = 0
gradient_steps = 0
training_history = []

# ------------------------------------------------------------
# 3. Initial evaluation with fresh episodes
# ------------------------------------------------------------

initial_eval = evaluate_dqn(
    agent,
    num_episodes=EVAL_EPISODES,
    seed=SEED + 300,
)

print("=" * 78)
print("IMPROVED DQN TRAINING")
print("=" * 78)
print(
    f"Initial greedy success: "
    f"{100 * initial_eval['success_rate']:.2f}%"
)
print("Controlled lookup baseline: 50.00%")
print("-" * 78)

# ------------------------------------------------------------
# 4. Training loop
# ------------------------------------------------------------

for batch_index in range(1, TRAIN_BATCHES + 1):

    observations = env.reset()

    episode_returns = np.zeros(NUM_ENVS, dtype=np.float32)
    batch_losses = []

    while not np.all(env.done):

        active = ~env.done
        active_indices = np.flatnonzero(active)

        epsilon = EPSILON_END + (
            EPSILON_START - EPSILON_END
        ) * max(
            0.0,
            1.0 - global_step / EPSILON_DECAY_STEPS
        )

        actions = agent.select_actions(
            observations, epsilon=epsilon
        )

        next_observations, rewards, dones, info = env.step(actions)

        # Only record transitions for environments that were active
        # before this step. Completed environments remain untouched.
        replay_buffer.add_batch(
            observations[active_indices],
            actions[active_indices],
            rewards[active_indices],
            next_observations[active_indices],
            dones[active_indices],
        )

        episode_returns[active_indices] += rewards[active_indices]
        observations = next_observations

        global_step += len(active_indices)

        # Use multiple updates for each vectorized environment step.
        if len(replay_buffer) >= max(BATCH_SIZE, LEARNING_STARTS):

            for _ in range(GRADIENT_UPDATES_PER_STEP):
                loss = agent.train_step(
                    replay_buffer,
                    batch_size=BATCH_SIZE,
                )

                if loss is not None:
                    batch_losses.append(loss)
                    gradient_steps += 1

                if (
                    gradient_steps > 0
                    and gradient_steps % TARGET_UPDATE_EVERY == 0
                ):
                    agent.update_target_network()

    batch_successes = int(env.episode_success.sum())

    record = {
        "batch": batch_index,
        "successes": batch_successes,
        "success_rate": batch_successes / NUM_ENVS,
        "mean_return": float(episode_returns.mean()),
        "mean_loss": (
            float(np.mean(batch_losses))
            if batch_losses else float("nan")
        ),
        "epsilon": float(epsilon),
        "global_step": global_step,
        "gradient_steps": gradient_steps,
        "replay_size": len(replay_buffer),
    }

    training_history.append(record)

    print(
        f"Batch {batch_index:02d}/{TRAIN_BATCHES} | "
        f"success={100 * record['success_rate']:6.2f}% | "
        f"return={record['mean_return']:+.3f} | "
        f"loss={record['mean_loss']:.5f} | "
        f"epsilon={epsilon:.3f} | "
        f"updates={gradient_steps}"
    )

# ------------------------------------------------------------
# 5. Final evaluation on fresh episodes
# ------------------------------------------------------------

final_eval = evaluate_dqn(
    agent,
    num_episodes=EVAL_EPISODES,
    seed=SEED + 400,
)

print()
print("=" * 78)
print("IMPROVED DQN TRAINING SUMMARY")
print("=" * 78)
print(
    f"Initial greedy success : "
    f"{100 * initial_eval['success_rate']:.2f}%"
)
print(
    f"Final greedy success   : "
    f"{100 * final_eval['success_rate']:.2f}% "
    f"({final_eval['successes']}/{final_eval['episodes']})"
)
print("Lookup baseline        : 50.00%")
print(f"Environment transitions: {global_step}")
print(f"Gradient updates       : {gradient_steps}")
print(f"Replay buffer size     : {len(replay_buffer)}")
print("=" * 78)
```

```python

# Cell 20: Isolated synthetic DQN learning test (corrected)

import numpy as np
import torch

print("=" * 70)
print("SYNTHETIC DQN LEARNING TEST")
print("=" * 70)

test_agent = DQNAgent(
    obs_dim=OBS_DIM,
    action_dim=NUM_ACTIONS,
    learning_rate=LEARNING_RATE,
    gamma=GAMMA,
    hidden_dim=HIDDEN_DIM,
    seed=SEED + 20,
)

test_buffer = ReplayBuffer(
    capacity=10_000,
    obs_dim=OBS_DIM,
    seed=SEED + 20,
)

rng = np.random.default_rng(SEED + 20)

NUM_SYNTHETIC = 2048

synthetic_states = rng.integers(
    0, 2, size=(NUM_SYNTHETIC, OBS_DIM)
).astype(np.float32)

synthetic_actions = rng.integers(
    0, NUM_ACTIONS, size=NUM_SYNTHETIC
).astype(np.int64)

synthetic_rewards = np.where(
    synthetic_actions == 0, 1.0, -1.0
).astype(np.float32)

synthetic_next_states = np.zeros(
    (NUM_SYNTHETIC, OBS_DIM), dtype=np.float32
)
synthetic_dones = np.ones(NUM_SYNTHETIC, dtype=bool)

test_buffer.add_batch(
    synthetic_states,
    synthetic_actions,
    synthetic_rewards,
    synthetic_next_states,
    synthetic_dones,
)

eval_states = rng.integers(
    0, 2, size=(512, OBS_DIM)
).astype(np.float32)


def synthetic_action_zero_rate(agent, states):
    agent.online_net.eval()

    # Infer the device from the network rather than agent.device.
    network_device = next(agent.online_net.parameters()).device

    with torch.no_grad():
        states_tensor = torch.as_tensor(
            states,
            dtype=torch.float32,
            device=network_device,
        )
        actions = agent.online_net(states_tensor).argmax(dim=1)

    agent.online_net.train()
    return (actions.cpu().numpy() == 0).mean()


initial_rate = synthetic_action_zero_rate(test_agent, eval_states)

NUM_UPDATES = 500
BATCH_SIZE_TEST = 128
test_losses = []

for update in range(NUM_UPDATES):
    loss = test_agent.train_step(
        test_buffer,
        batch_size=BATCH_SIZE_TEST,
    )
    test_losses.append(loss)

final_rate = synthetic_action_zero_rate(test_agent, eval_states)

print(f"Device                         : {next(test_agent.online_net.parameters()).device}")
print(f"Synthetic transitions          : {len(test_buffer)}")
print(f"Gradient updates               : {NUM_UPDATES}")
print(f"Initial action-0 selection rate: {initial_rate:.2%}")
print(f"Final action-0 selection rate  : {final_rate:.2%}")
print(f"Final training loss            : {np.mean(test_losses[-20:]):.5f}")

if final_rate >= 0.95:
    print("\nPASS: The network learned the synthetic reward rule.")
else:
    print(
        "\nNOT YET PASSED: The network did not reliably learn "
        "the synthetic rule. We should diagnose this before "
        "resuming QEC training."
    )
```

```python

# Cell 21: QEC reward and single-action diagnostic

import numpy as np

print("=" * 72)
print("QEC REWARD AND ACTION DIAGNOSTIC")
print("=" * 72)

# Validate the required arrays.
required_arrays = {
    "error_patterns": error_patterns,
    "detector_syndromes": detector_syndromes,
    "logical_observables": logical_observables,
    "single_detector_effects": single_detector_effects,
    "single_observable_effects": single_observable_effects,
}

for name, array in required_arrays.items():
    print(f"{name:30s}: {np.asarray(array).shape}")

N = len(error_patterns)

assert detector_syndromes.shape[0] == N
assert logical_observables.shape[0] == N
assert single_detector_effects.shape[0] == NUM_DATA_QUBITS
assert single_observable_effects.shape[0] == NUM_DATA_QUBITS

# Evaluate a correction frame against every dataset pattern.
def evaluate_correction_frame(correction_frame):
    correction_frame = np.asarray(correction_frame, dtype=np.int64)

    correction_syndrome = (
        correction_frame @ single_detector_effects
    ) % 2

    correction_logical = (
        correction_frame @ single_observable_effects
    ) % 2

    residual_syndromes = (
        detector_syndromes ^ correction_syndrome
    )

    residual_logical = (
        logical_observables ^ correction_logical
    )

    syndrome_ok = np.all(residual_syndromes == 0, axis=1)
    logical_ok = np.all(residual_logical == 0, axis=1)
    success = syndrome_ok & logical_ok

    return {
        "success": success,
        "syndrome_ok": syndrome_ok,
        "logical_ok": logical_ok,
    }


# Baseline: stop without applying a correction.
zero_frame = np.zeros(NUM_DATA_QUBITS, dtype=np.int64)
baseline = evaluate_correction_frame(zero_frame)

print("\nBASELINE: STOP WITHOUT CORRECTION")
print(f"Syndrome cleared : {baseline['syndrome_ok'].sum()}/{N} "
      f"({baseline['syndrome_ok'].mean():.2%})")
print(f"Logical correct  : {baseline['logical_ok'].sum()}/{N} "
      f"({baseline['logical_ok'].mean():.2%})")
print(f"Overall success  : {baseline['success'].sum()}/{N} "
      f"({baseline['success'].mean():.2%})")

# Evaluate one data-qubit correction followed by STOP.
print("\nSINGLE-QUBIT CORRECTION RESULTS")
print("-" * 72)
print(f"{'Action':>8} {'Syndrome cleared':>20} "
      f"{'Logical correct':>18} {'Overall success':>18}")

single_action_results = {}

for q in range(NUM_DATA_QUBITS):
    frame = np.zeros(NUM_DATA_QUBITS, dtype=np.int64)
    frame[q] = 1

    result = evaluate_correction_frame(frame)
    single_action_results[q] = result

    print(
        f"{q:>8} "
        f"{result['syndrome_ok'].mean():>19.2%} "
        f"{result['logical_ok'].mean():>17.2%} "
        f"{result['success'].mean():>17.2%}"
    )

# STOP action means keep the current zero correction frame.
print(
    f"{STOP_ACTION:>8} "
    f"{baseline['syndrome_ok'].mean():>19.2%} "
    f"{baseline['logical_ok'].mean():>17.2%} "
    f"{baseline['success'].mean():>17.2%}  (STOP)"
)

# Sanity check: applying the actual injected error pattern
# as the correction must clear its syndrome and logical effect.
oracle_results = []

for pattern in error_patterns:
    result = evaluate_correction_frame(pattern)
    oracle_results.append(result["success"][len(oracle_results)])

oracle_success = np.asarray(oracle_results, dtype=bool)

print("\nORACLE SANITY CHECK")
print(f"Success with exact error-pattern correction: "
      f"{oracle_success.sum()}/{N} ({oracle_success.mean():.2%})")

assert oracle_success.all(), (
    "Oracle correction failed. Investigate the effect arrays "
    "before continuing."
)

print("\nDiagnostic completed.")
```

```python

# Cell 22: Verify the lookup decoder inside the actual RL environment

import numpy as np

print("=" * 72)
print("LOOKUP DECODER ENVIRONMENT TEST")
print("=" * 72)

# Build a minimum-weight correction for each observed syndrome.
lookup = {}

for pattern, syndrome in zip(error_patterns, detector_syndromes):
    key = tuple(np.asarray(syndrome, dtype=np.int8))
    weight = int(np.sum(pattern))

    if key not in lookup or weight < lookup[key][0]:
        lookup[key] = (weight, np.asarray(pattern, dtype=np.int8).copy())

print(f"Unique syndrome entries in lookup: {len(lookup)}")

# Create an independent environment with one slot per evaluation episode.
N_EVAL = 512

oracle_env = VectorizedQECEnv(
    controlled_dataset,
    single_detector_effects,
    single_observable_effects,
    num_envs=N_EVAL,
    max_steps=10,
    stop_action=STOP_ACTION,
    seed=SEED + 22,
)

reset_result = oracle_env.reset()
observations = (
    reset_result[0]
    if isinstance(reset_result, tuple)
    else reset_result
)
observations = np.asarray(observations, dtype=np.float32)

n_detectors = detector_syndromes.shape[1]

# Select the minimum-weight correction from the observed syndrome.
target_corrections = np.zeros(
    (N_EVAL, NUM_DATA_QUBITS), dtype=np.int8
)

for i in range(N_EVAL):
    syndrome_key = tuple(
        observations[i, :n_detectors].astype(np.int8)
    )
    target_corrections[i] = lookup[syndrome_key][1]

# Each active environment applies one remaining correction bit per step.
remaining = target_corrections.copy()
total_rewards = np.zeros(N_EVAL, dtype=np.float32)
finished = np.zeros(N_EVAL, dtype=bool)

def unpack_step(result):
    if len(result) == 5:
        next_obs, rewards, terminated, truncated, info = result
        dones = np.asarray(terminated) | np.asarray(truncated)
    elif len(result) == 4:
        next_obs, rewards, dones, info = result
    elif len(result) == 3:
        next_obs, rewards, dones = result
    else:
        raise ValueError(
            f"Unexpected environment step result length: {len(result)}"
        )

    return (
        np.asarray(next_obs),
        np.asarray(rewards, dtype=np.float32).reshape(-1),
        np.asarray(dones, dtype=bool).reshape(-1),
    )

for step_index in range(10):
    actions = np.full(N_EVAL, STOP_ACTION, dtype=np.int64)

    for i in range(N_EVAL):
        if finished[i]:
            continue

        remaining_qubits = np.flatnonzero(remaining[i])

        if len(remaining_qubits) > 0:
            q = int(remaining_qubits[0])
            actions[i] = q
            remaining[i, q] = 0
        # If no correction bits remain, STOP is selected.

    observations, rewards, dones = unpack_step(
        oracle_env.step(actions)
    )

    total_rewards += rewards
    finished |= dones

    if finished.all():
        break

print(f"Episodes evaluated : {N_EVAL}")
print(f"Episodes terminated: {finished.sum()}/{N_EVAL}")
print(f"Successful episodes: {(total_rewards > 0).sum()}/{N_EVAL}")
print(f"Environment success: {(total_rewards > 0).mean():.2%}")
print(f"Mean episode reward: {total_rewards.mean():.4f}")

if not finished.all():
    print(
        "\nWARNING: Some episodes did not terminate within the "
        "expected action budget. Inspect the environment API."
    )
elif 0.35 <= (total_rewards > 0).mean() <= 0.65:
    print(
        "\nPASS: The environment accepts lookup corrections and "
        "achieves a success rate consistent with the 50% "
        "controlled-dataset baseline."
    )
else:
    print(
        "\nINVESTIGATE: The measured rate differs substantially "
        "from the expected 50%. Check the output and environment "
        "behavior before changing DQN training."
    )
```

```python

# Cell 23: Inspect trained DQN action preferences

import numpy as np
import torch

print("=" * 72)
print("TRAINED DQN ACTION-PREFERENCE DIAGNOSTIC")
print("=" * 72)

# Use a separate evaluation environment.
N_DIAGNOSTIC = 512

diagnostic_env = VectorizedQECEnv(
    controlled_dataset,
    single_detector_effects,
    single_observable_effects,
    num_envs=N_DIAGNOSTIC,
    max_steps=10,
    stop_action=STOP_ACTION,
    seed=SEED + 23,
)

reset_result = diagnostic_env.reset()
diagnostic_obs = (
    reset_result[0]
    if isinstance(reset_result, tuple)
    else reset_result
)
diagnostic_obs = np.asarray(diagnostic_obs, dtype=np.float32)

# Use the original trained agent, not the synthetic test agent.
trained_agent = agent if "agent" in globals() else dqn_agent

trained_agent.online_net.eval()
network_device = next(
    trained_agent.online_net.parameters()
).device

with torch.no_grad():
    obs_tensor = torch.as_tensor(
        diagnostic_obs,
        dtype=torch.float32,
        device=network_device,
    )
    q_values = trained_agent.online_net(obs_tensor).cpu().numpy()

trained_agent.online_net.train()

greedy_actions = np.argmax(q_values, axis=1)
action_counts = np.bincount(
    greedy_actions,
    minlength=NUM_ACTIONS,
)

print(f"Observations evaluated: {len(diagnostic_obs)}")
print(f"Q-value matrix shape  : {q_values.shape}")
print(f"Q-value mean          : {q_values.mean():.5f}")
print(f"Q-value standard dev. : {q_values.std():.5f}")
print(f"Q-value minimum       : {q_values.min():.5f}")
print(f"Q-value maximum       : {q_values.max():.5f}")

print("\nGREEDY ACTION FREQUENCIES")
print("-" * 48)

for action, count in enumerate(action_counts):
    label = "STOP" if action == STOP_ACTION else f"Correct qubit {action}"
    print(f"{label:20s}: {count:4d} ({count / N_DIAGNOSTIC:.2%})")

print("\nACTION VALUE SUMMARY")
print("-" * 48)

mean_q_by_action = q_values.mean(axis=0)

for action, mean_q in enumerate(mean_q_by_action):
    label = "STOP" if action == STOP_ACTION else f"Qubit {action}"
    print(f"{label:20s}: mean Q = {mean_q:.5f}")

print("\nDiagnostic completed; no training updates were performed.")
```

```python

# Cell 24: Evaluate the trained policy on every unique syndrome

import numpy as np
import torch

print("=" * 76)
print("DQN POLICY BY UNIQUE SYNDROME")
print("=" * 76)

trained_agent = agent if "agent" in globals() else dqn_agent

# Build one initial observation for each unique syndrome.
unique_syndromes = np.unique(detector_syndromes, axis=0)
zero_corrections = np.zeros(
    (len(unique_syndromes), NUM_DATA_QUBITS),
    dtype=np.float32,
)

syndrome_observations = np.concatenate(
    [
        unique_syndromes.astype(np.float32),
        zero_corrections,
    ],
    axis=1,
)

network_device = next(
    trained_agent.online_net.parameters()
).device

trained_agent.online_net.eval()

with torch.no_grad():
    obs_tensor = torch.as_tensor(
        syndrome_observations,
        dtype=torch.float32,
        device=network_device,
    )
    syndrome_q_values = (
        trained_agent.online_net(obs_tensor).cpu().numpy()
    )

trained_agent.online_net.train()

greedy_actions = syndrome_q_values.argmax(axis=1)

print(f"Unique syndromes tested: {len(unique_syndromes)}")
print(f"Observation shape      : {syndrome_observations.shape}")
print()
print(
    f"{'ID':>3} {'Syndrome weight':>16} "
    f"{'Greedy action':>15} {'Best Q':>10} {'STOP Q':>10}"
)
print("-" * 62)

for i, syndrome in enumerate(unique_syndromes):
    action = int(greedy_actions[i])
    action_name = (
        "STOP" if action == STOP_ACTION
        else f"Qubit {action}"
    )

    print(
        f"{i:>3} "
        f"{int(syndrome.sum()):>16} "
        f"{action_name:>15} "
        f"{syndrome_q_values[i, action]:>10.4f} "
        f"{syndrome_q_values[i, STOP_ACTION]:>10.4f}"
    )

print("\nGREEDY ACTION COUNTS ACROSS UNIQUE SYNDROMES")
for action, count in enumerate(
    np.bincount(greedy_actions, minlength=NUM_ACTIONS)
):
    label = "STOP" if action == STOP_ACTION else f"Qubit {action}"
    print(f"{label:12s}: {count:2d}/{len(unique_syndromes)}")

print("\nPolicy-by-syndrome diagnostic completed.")
```

```python

# Cell 25: Supervised imitation diagnostic using a separate network

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim

print("=" * 72)
print("SUPERVISED POLICY IMITATION DIAGNOSTIC")
print("=" * 72)

# Rebuild deterministic minimum-weight corrections.
lookup_targets = {}

for pattern, syndrome in zip(error_patterns, detector_syndromes):
    key = tuple(np.asarray(syndrome, dtype=np.int8))
    weight = int(np.sum(pattern))

    if key not in lookup_targets or weight < lookup_targets[key][0]:
        lookup_targets[key] = (
            weight,
            np.asarray(pattern, dtype=np.int8).copy(),
        )

# Generate observations and target actions along each correction sequence.
training_states = []
training_actions = []

for syndrome_key, (_, target_frame) in lookup_targets.items():
    syndrome = np.asarray(syndrome_key, dtype=np.float32)
    current_frame = np.zeros(NUM_DATA_QUBITS, dtype=np.int8)

    # The lookup decoder applies each required correction once.
    for q in np.flatnonzero(target_frame):
        state = np.concatenate(
            [syndrome, current_frame.astype(np.float32)]
        )
        training_states.append(state)
        training_actions.append(int(q))

        current_frame[q] ^= 1

    # Once the target frame is reached, stop.
    final_state = np.concatenate(
        [syndrome, current_frame.astype(np.float32)]
    )
    training_states.append(final_state)
    training_actions.append(STOP_ACTION)

X = np.asarray(training_states, dtype=np.float32)
y = np.asarray(training_actions, dtype=np.int64)

print(f"Training states       : {X.shape}")
print(f"Target action labels  : {y.shape}")
print(f"Unique syndromes      : {len(lookup_targets)}")
print(f"Target action classes : {np.unique(y).tolist()}")

# Train a fresh network; the original RL agent remains untouched.
torch.manual_seed(SEED + 25)
np.random.seed(SEED + 25)

imitation_net = QNetwork(
    obs_dim=OBS_DIM,
    action_dim=NUM_ACTIONS,
    hidden_dim=HIDDEN_DIM,
).to(next(trained_agent.online_net.parameters()).device)

optimizer = optim.Adam(imitation_net.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

X_tensor = torch.as_tensor(
    X,
    dtype=torch.float32,
    device=next(imitation_net.parameters()).device,
)
y_tensor = torch.as_tensor(
    y,
    dtype=torch.long,
    device=next(imitation_net.parameters()).device,
)

# Fixed, small supervised diagnostic.
EPOCHS = 400
for epoch in range(EPOCHS):
    imitation_net.train()
    optimizer.zero_grad()

    logits = imitation_net(X_tensor)
    loss = criterion(logits, y_tensor)

    loss.backward()
    optimizer.step()

imitation_net.eval()
with torch.no_grad():
    predictions = imitation_net(X_tensor).argmax(dim=1)
    accuracy = (predictions == y_tensor).float().mean().item()

print(f"\nTraining epochs      : {EPOCHS}")
print(f"Final imitation loss : {loss.item():.6f}")
print(f"Training action accuracy: {accuracy:.2%}")

if accuracy >= 0.95:
    print(
        "\nPASS: The network can represent and learn the lookup "
        "correction sequences with supervised targets."
    )
else:
    print(
        "\nNOT YET PASSED: The imitation task was not learned "
        "reliably. Inspect architecture and target construction."
    )

print("\nOriginal RL agent weights were not modified.")
```

```python

# Cell 26: First-action generalization on all controlled patterns

import numpy as np
import torch

print("=" * 72)
print("IMITATION POLICY GENERALIZATION TEST")
print("=" * 72)

# Use the actual syndrome of every controlled error pattern.
evaluation_states = np.concatenate(
    [
        detector_syndromes.astype(np.float32),
        np.zeros(
            (len(detector_syndromes), NUM_DATA_QUBITS),
            dtype=np.float32,
        ),
    ],
    axis=1,
)

# Determine the first lookup action for each syndrome.
lookup_first_action = {}

for pattern, syndrome in zip(error_patterns, detector_syndromes):
    key = tuple(np.asarray(syndrome, dtype=np.int8))
    weight = int(np.sum(pattern))

    if (
        key not in lookup_first_action
        or weight < lookup_first_action[key][0]
    ):
        nonzero = np.flatnonzero(pattern)
        first_action = int(nonzero[0]) if len(nonzero) else STOP_ACTION
        lookup_first_action[key] = (weight, first_action)

target_actions = np.asarray(
    [
        lookup_first_action[
            tuple(np.asarray(s, dtype=np.int8))
        ][1]
        for s in detector_syndromes
    ],
    dtype=np.int64,
)

network_device = next(imitation_net.parameters()).device
imitation_net.eval()

with torch.no_grad():
    states_tensor = torch.as_tensor(
        evaluation_states,
        dtype=torch.float32,
        device=network_device,
    )
    predicted_actions = (
        imitation_net(states_tensor).argmax(dim=1).cpu().numpy()
    )

accuracy = np.mean(predicted_actions == target_actions)

print(f"Evaluation patterns       : {len(evaluation_states)}")
print(f"Unique syndromes          : {len(unique_syndromes)}")
print(f"First-action agreement    : {accuracy:.2%}")

print("\nPredicted action distribution:")
for action, count in enumerate(
    np.bincount(predicted_actions, minlength=NUM_ACTIONS)
):
    label = "STOP" if action == STOP_ACTION else f"Qubit {action}"
    print(f"{label:12s}: {count:3d} ({count / len(predicted_actions):.2%})")

print("\nLookup action distribution:")
for action, count in enumerate(
    np.bincount(target_actions, minlength=NUM_ACTIONS)
):
    label = "STOP" if action == STOP_ACTION else f"Qubit {action}"
    print(f"{label:12s}: {count:3d} ({count / len(target_actions):.2%})")

print("\nGeneralization test completed.")
```

```python

# Cell 27: Full imitation-policy evaluation inside the QEC environment

import numpy as np
import torch

print("=" * 72)
print("FULL IMITATION-POLICY ENVIRONMENT EVALUATION")
print("=" * 72)

N_EVAL = 512

eval_env = VectorizedQECEnv(
    controlled_dataset,
    single_detector_effects,
    single_observable_effects,
    num_envs=N_EVAL,
    max_steps=10,
    stop_action=STOP_ACTION,
    seed=SEED + 27,
)

reset_result = eval_env.reset()
observations = (
    reset_result[0]
    if isinstance(reset_result, tuple)
    else reset_result
)
observations = np.asarray(observations, dtype=np.float32)

episode_rewards = np.zeros(N_EVAL, dtype=np.float32)
finished = np.zeros(N_EVAL, dtype=bool)
action_counts = np.zeros(NUM_ACTIONS, dtype=np.int64)

network_device = next(imitation_net.parameters()).device
imitation_net.eval()

for step_index in range(10):
    # Greedy action selection from the supervised imitation network.
    with torch.no_grad():
        obs_tensor = torch.as_tensor(
            observations,
            dtype=torch.float32,
            device=network_device,
        )
        actions = (
            imitation_net(obs_tensor)
            .argmax(dim=1)
            .cpu()
            .numpy()
            .astype(np.int64)
        )

    # Count actions only for episodes that were active before this step.
    for action in actions[~finished]:
        action_counts[action] += 1

    step_result = eval_env.step(actions)

    if len(step_result) == 5:
        next_obs, rewards, terminated, truncated, info = step_result
        dones = np.asarray(terminated) | np.asarray(truncated)
    elif len(step_result) == 4:
        next_obs, rewards, dones, info = step_result
    elif len(step_result) == 3:
        next_obs, rewards, dones = step_result
    else:
        raise ValueError(
            f"Unexpected step result length: {len(step_result)}"
        )

    rewards = np.asarray(rewards, dtype=np.float32).reshape(-1)
    dones = np.asarray(dones, dtype=bool).reshape(-1)

    episode_rewards += rewards
    finished |= dones
    observations = np.asarray(next_obs, dtype=np.float32)

    if finished.all():
        break

imitation_net.train()

successes = int(np.sum(episode_rewards > 0))

print(f"Episodes evaluated : {N_EVAL}")
print(f"Episodes terminated: {finished.sum()}/{N_EVAL}")
print(f"Successful episodes: {successes}/{N_EVAL}")
print(f"Success rate       : {successes / N_EVAL:.2%}")
print(f"Mean episode reward: {episode_rewards.mean():.4f}")

print("\nActions taken during evaluation:")
for action, count in enumerate(action_counts):
    label = "STOP" if action == STOP_ACTION else f"Qubit {action}"
    print(f"{label:12s}: {count:5d}")

print("\nReference controlled lookup baseline: approximately 50%")

if not finished.all():
    print("WARNING: Some episodes did not terminate within 10 steps.")
elif successes / N_EVAL >= 0.40:
    print("PASS: Full imitation policy is consistent with the baseline.")
else:
    print("INVESTIGATE: Full-sequence performance is below expectation.")
```

```python
# Cell 33: Diagnose STOP behaviour, episode lengths, and success
import numpy as np

DIAG_EPISODES = 2048
DIAG_SEED = SEED + 500

diag_env = VectorizedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=DIAG_EPISODES,
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=DIAG_SEED,
)

obs = diag_env.reset()

# Count only actions taken by environments that are still active.
action_counts = np.zeros(ACTION_DIM, dtype=np.int64)
stop_step_counts = np.zeros(MAX_STEPS + 1, dtype=np.int64)
actions_by_step = np.zeros((MAX_STEPS, ACTION_DIM), dtype=np.int64)

while not np.all(diag_env.done):
    active_before = ~diag_env.done
    active_indices = np.flatnonzero(active_before)

    actions = agent.select_actions(obs, epsilon=0.0)
    active_actions = actions[active_indices]
    current_steps = diag_env.steps[active_indices] + 1

    for action in range(ACTION_DIM):
        action_counts[action] += np.count_nonzero(
            active_actions == action
        )

    for step_number in range(1, MAX_STEPS + 1):
        at_step = current_steps == step_number
        if np.any(at_step):
            for action in range(ACTION_DIM):
                actions_by_step[step_number - 1, action] += (
                    np.count_nonzero(active_actions[at_step] == action)
                )

    stop_mask = active_actions == STOP_ACTION
    if np.any(stop_mask):
        stop_step_counts[current_steps[stop_mask]] += 1

    obs, rewards, dones, info = diag_env.step(actions)

# Summarize terminal outcomes.
episode_lengths = diag_env.steps
successes = int(diag_env.episode_success.sum())
stop_episodes = int(stop_step_counts.sum())
timeout_episodes = int(DIAG_EPISODES - stop_episodes)

print("=" * 72)
print("GREEDY DQN POLICY DIAGNOSTICS")
print("=" * 72)
print(f"Episodes             : {DIAG_EPISODES}")
print(f"Successes            : {successes}")
print(f"Success rate         : {100 * successes / DIAG_EPISODES:.2f}%")
print(f"STOP-terminated      : {stop_episodes} "
      f"({100 * stop_episodes / DIAG_EPISODES:.2f}%)")
print(f"Timed out            : {timeout_episodes} "
      f"({100 * timeout_episodes / DIAG_EPISODES:.2f}%)")
print(f"Mean episode length  : {episode_lengths.mean():.2f}")
print(f"Median episode length: {np.median(episode_lengths):.1f}")

print("\nAction counts across active steps:")
for action, count in enumerate(action_counts):
    label = "STOP" if action == STOP_ACTION else f"Toggle correction {action}"
    print(f"  {action:2d} ({label:22s}): {count:7d}")

print("\nAction counts by step:")
print("Step | Active actions | STOP | Most frequent action")
print("-" * 58)

for step_number in range(1, MAX_STEPS + 1):
    counts = actions_by_step[step_number - 1]
    total = int(counts.sum())
    if total == 0:
        continue

    most_common = int(np.argmax(counts))
    label = "STOP" if most_common == STOP_ACTION else f"Toggle {most_common}"

    print(
        f"{step_number:4d} | {total:14d} | "
        f"{counts[STOP_ACTION]:4d} | {label}"
    )

print("\nDiagnostic complete. The trained agent was not updated.")
```

```python
class TimedVectorizedQECEnv(VectorizedQECEnv):
    """Experimental environment with remaining time in its observation."""

    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.observation_dim = (
            self.num_detectors + self.num_data_qubits + 1
        )

    def _get_observation(self):
        remaining_fraction = (
            (self.max_steps - self.steps) / self.max_steps
        ).astype(np.float32).reshape(-1, 1)

        return np.concatenate(
            [self.detectors, self.corrections, remaining_fraction],
            axis=1,
        ).astype(np.float32)


# Reuse the original dataset and noise-effect matrices.
timed_test_env = TimedVectorizedQECEnv(
    dataset=env.dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=4,
    max_steps=env.max_steps,
    stop_action=env.stop_action,
    seed=42,
)

obs = timed_test_env.reset()

print("Observation shape:", obs.shape)
print("Declared observation dimension:", timed_test_env.observation_dim)
print("Initial remaining-step fractions:", obs[:, -1])

# Take one correction-toggle action in each test environment.
actions = np.zeros(4, dtype=np.int64)
next_obs, rewards, dones, info = timed_test_env.step(actions)

print("Remaining-step fractions after one step:", next_obs[:, -1])
print("Rewards:", rewards)
print("Done flags:", dones)

assert obs.shape == (4, 34)
assert np.allclose(obs[:, -1], 1.0)
assert np.allclose(next_obs[:, -1], 0.9)

print("\nTimed observation test passed.")
```

```python

def residual_syndrome_weight(qec_env):
    """Count residual syndrome bits for every environment."""
    correction_syndromes = (
        qec_env.corrections @ qec_env.single_detector_effects
    ) % 2

    residual = qec_env.detectors ^ correction_syndromes
    return np.sum(residual, axis=1).astype(np.float32)


def syndrome_potential(qec_env):
    """Potential: negative residual syndrome weight."""
    return -residual_syndrome_weight(qec_env)


def potential_shaping_reward(phi_before, phi_after, gamma=0.99):
    """Potential-based shaping term for a transition."""
    return gamma * phi_after - phi_before


# Use a fresh reset of the experimental environment.
obs = timed_test_env.reset()

phi_before = syndrome_potential(timed_test_env)

# Apply one toggle action to each environment.
actions = np.zeros(timed_test_env.num_envs, dtype=np.int64)
next_obs, terminal_rewards, dones, info = timed_test_env.step(actions)

phi_after = syndrome_potential(timed_test_env)
shaping = potential_shaping_reward(phi_before, phi_after, gamma=0.99)

print("Initial residual syndrome weights:",
      -phi_before)
print("New residual syndrome weights:",
      -phi_after)
print("Original environment rewards:", terminal_rewards)
print("Potential-shaping rewards:", shaping)

assert phi_before.shape == (timed_test_env.num_envs,)
assert phi_after.shape == (timed_test_env.num_envs,)
assert shaping.shape == (timed_test_env.num_envs,)
assert np.all(np.isfinite(shaping))

print("\nReward-shaping calculation passed.")
```

```python

class ShapedTimedQECEnv(TimedVectorizedQECEnv):
    """Timed QEC environment with potential-based reward shaping."""

    def __init__(self, *args, shaping_gamma=0.99, **kwargs):
        super().__init__(*args, **kwargs)
        self.shaping_gamma = shaping_gamma

    def step(self, actions):
        # Potential before the transition.
        phi_before = syndrome_potential(self)

        next_obs, original_rewards, dones, info = super().step(actions)

        # Potential after the transition; terminal potential is zero.
        phi_after = syndrome_potential(self)
        phi_after = np.where(dones, 0.0, phi_after)

        shaping = (
            self.shaping_gamma * phi_after - phi_before
        )

        # Do not shape environments that were already done before this step.
        # Their previous potential is irrelevant because they did not transition.
        rewards = original_rewards + shaping.astype(np.float32)

        return next_obs, rewards, dones, info


shaped_test_env = ShapedTimedQECEnv(
    dataset=env.dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=4,
    max_steps=env.max_steps,
    stop_action=env.stop_action,
    seed=42,
    shaping_gamma=0.99,
)

obs = shaped_test_env.reset()
old_potential = syndrome_potential(shaped_test_env)

actions = np.zeros(4, dtype=np.int64)
next_obs, rewards, dones, info = shaped_test_env.step(actions)

new_potential = syndrome_potential(shaped_test_env)
expected_shaping = 0.99 * new_potential - old_potential

print("Rewards returned by shaped environment:", rewards)
print("Expected shaping-only rewards:", expected_shaping)
print("Done flags:", dones)

assert np.allclose(rewards, expected_shaping)
assert np.all(np.isfinite(rewards))

print("\nShaped environment test passed.")
```

```python

class ShapedTimedQECEnv(TimedVectorizedQECEnv):
    """Timed QEC environment with correctly masked potential shaping."""

    def __init__(self, *args, shaping_gamma=0.99, **kwargs):
        super().__init__(*args, **kwargs)
        self.shaping_gamma = shaping_gamma

    def step(self, actions):
        # Record which environments actually take a transition.
        active_before = ~self.done

        phi_before = syndrome_potential(self)

        next_obs, original_rewards, dones, info = super().step(actions)

        phi_after = syndrome_potential(self)

        # Terminal states have zero potential.
        phi_after = np.where(dones, 0.0, phi_after)

        shaping = self.shaping_gamma * phi_after - phi_before

        # Already-finished environments must receive no additional reward.
        shaping = np.where(active_before, shaping, 0.0)

        rewards = original_rewards + shaping.astype(np.float32)

        return next_obs, rewards, dones, info


# Create matched environments using the same dataset and random seed.
base_check_env = TimedVectorizedQECEnv(
    dataset=env.dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=4,
    max_steps=env.max_steps,
    stop_action=env.stop_action,
    seed=123,
)

shaped_check_env = ShapedTimedQECEnv(
    dataset=env.dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=4,
    max_steps=env.max_steps,
    stop_action=env.stop_action,
    seed=123,
    shaping_gamma=0.99,
)

base_check_env.reset()
shaped_check_env.reset()

# Confirm both environments sampled the same hidden cases.
assert np.array_equal(
    base_check_env.error_indices,
    shaped_check_env.error_indices,
)

phi_before = syndrome_potential(shaped_check_env)

# STOP immediately, forcing terminal transitions.
stop_actions = np.full(4, shaped_check_env.stop_action, dtype=np.int64)

_, base_rewards, base_dones, _ = base_check_env.step(stop_actions)
_, shaped_rewards, shaped_dones, _ = shaped_check_env.step(stop_actions)

expected_rewards = base_rewards - phi_before

print("Base terminal rewards:", base_rewards)
print("Initial residual syndrome weights:", -phi_before)
print("Shaped terminal rewards:", shaped_rewards)
print("Expected shaped rewards:", expected_rewards)
print("Terminal flags:", shaped_dones)

assert np.all(base_dones)
assert np.all(shaped_dones)
assert np.allclose(shaped_rewards, expected_rewards)

# Calling step again must not reward already-finished environments.
_, inactive_rewards, inactive_dones, _ = shaped_check_env.step(stop_actions)

print("Rewards after episodes already ended:", inactive_rewards)

assert np.allclose(inactive_rewards, 0.0)
assert np.all(inactive_dones)

print("\nTerminal and inactive-environment tests passed.")
```

```python

def evaluate_shaped_dqn(agent, num_episodes=2048, seed=300):
    """Evaluate a timed, shaped DQN on fresh episodes."""
    eval_env = ShapedTimedQECEnv(
        dataset=controlled_dataset,
        single_detector_effects=single_detector_effects,
        single_observable_effects=single_observable_effects,
        num_envs=num_episodes,
        max_steps=MAX_STEPS,
        stop_action=STOP_ACTION,
        seed=seed,
        shaping_gamma=GAMMA,
    )

    observations = eval_env.reset()

    while not np.all(eval_env.done):
        actions = agent.select_actions(observations, epsilon=0.0)
        observations, _, _, _ = eval_env.step(actions)

    successes = int(eval_env.episode_success.sum())

    return {
        "successes": successes,
        "episodes": num_episodes,
        "success_rate": successes / num_episodes,
    }


# Evaluate before any meaningful training.
shaped_initial_eval = evaluate_shaped_dqn(
    shaped_agent,
    num_episodes=EVAL_EPISODES,
    seed=SEED + 300,
)

print("Experimental observation dimension:", shaped_agent.obs_dim)
print(
    f"Initial greedy success: "
    f"{100 * shaped_initial_eval['success_rate']:.2f}% "
    f"({shaped_initial_eval['successes']}/"
    f"{shaped_initial_eval['episodes']})"
)
print("Lookup baseline: approximately 50%")

assert shaped_initial_eval["episodes"] == EVAL_EPISODES

print("\nExperimental evaluator passed.")
```

```python
import numpy as np

shaped_training_history = []
shaped_global_step = 0
shaped_gradient_steps = 0
shaped_last_target_update = 0

for batch_idx in range(TRAIN_BATCHES):
    observations = shaped_train_env.reset()
    episode_returns = np.zeros(NUM_ENVS, dtype=np.float32)
    batch_losses = []

    while not np.all(shaped_train_env.done):
        # Only environments that have not finished contribute transitions.
        active = ~shaped_train_env.done
        active_indices = np.flatnonzero(active)

        epsilon = EPSILON_END + (
            EPSILON_START - EPSILON_END
        ) * max(
            0.0,
            1.0 - shaped_global_step / EPSILON_DECAY_STEPS
        )

        actions = shaped_agent.select_actions(
            observations,
            epsilon=epsilon,
        )

        next_observations, rewards, dones, info = shaped_train_env.step(
            actions
        )

        shaped_replay_buffer.add_batch(
            observations[active_indices],
            actions[active_indices],
            rewards[active_indices],
            next_observations[active_indices],
            dones[active_indices],
        )

        episode_returns[active_indices] += rewards[active_indices]
        observations = next_observations
        shaped_global_step += len(active_indices)

        # Gradient updates, matching the original experiment's schedule.
        if shaped_replay_buffer.size >= max(BATCH_SIZE, LEARNING_STARTS):
            for _ in range(GRADIENT_UPDATES_PER_STEP):
                loss = shaped_agent.train_step(
                    shaped_replay_buffer,
                    batch_size=BATCH_SIZE,
                )
                batch_losses.append(float(loss))
                shaped_gradient_steps += 1

                # Periodically synchronize the target network.
                if shaped_gradient_steps % TARGET_UPDATE_EVERY == 0:
                    shaped_agent.target_net.load_state_dict(
                        shaped_agent.q_net.state_dict()
                    )
                    shaped_last_target_update = shaped_gradient_steps

    batch_successes = int(shaped_train_env.episode_success.sum())
    batch_success_rate = batch_successes / NUM_ENVS
    mean_shaped_return = float(episode_returns.mean())
    mean_loss = (
        float(np.mean(batch_losses)) if batch_losses else float("nan")
    )

    batch_record = {
        "batch": batch_idx + 1,
        "successes": batch_successes,
        "success_rate": batch_success_rate,
        "mean_shaped_return": mean_shaped_return,
        "mean_loss": mean_loss,
        "epsilon": epsilon,
        "global_steps": shaped_global_step,
        "gradient_steps": shaped_gradient_steps,
    }
    shaped_training_history.append(batch_record)

    print(
        f"Batch {batch_idx + 1:02d}/{TRAIN_BATCHES} | "
        f"Success: {batch_success_rate * 100:6.2f}% "
        f"({batch_successes}/{NUM_ENVS}) | "
        f"Mean shaped return: {mean_shaped_return:7.3f} | "
        f"Loss: {mean_loss:.5f} | "
        f"Epsilon: {epsilon:.3f} | "
        f"Gradients: {shaped_gradient_steps}"
    )

print("\nTraining finished.")
print("Environment transitions:", shaped_global_step)
print("Gradient updates:", shaped_gradient_steps)
print("Replay-buffer size:", shaped_replay_buffer.size)
print("Last target-network update:", shaped_last_target_update)
```

```python
# Diagnose STOP selection during greedy DQN evaluation

action_eval_env = ShapedTimedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=len(controlled_dataset["error_patterns"]),
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=123,
    shaping_gamma=GAMMA,
)

obs = action_eval_env.reset()

# Evaluate every dataset pattern exactly once.
action_eval_env.error_indices = np.arange(len(controlled_dataset["error_patterns"]))
action_eval_env.hidden_errors = controlled_dataset["error_patterns"].copy()
action_eval_env.hidden_logical = controlled_dataset["logical_observables"].copy()
action_eval_env.detectors = controlled_dataset["detectors"].copy()
action_eval_env.corrections.fill(0)
action_eval_env.steps.fill(0)
action_eval_env.done.fill(False)
action_eval_env.episode_success.fill(False)
obs = action_eval_env._get_observation()

action_counts = np.zeros(NUM_ACTIONS, dtype=np.int64)
stop_step = np.full(action_eval_env.num_envs, -1, dtype=np.int64)

while not np.all(action_eval_env.done):
    active_before = ~action_eval_env.done
    actions = shaped_agent.select_actions(obs, epsilon=0.0)

    # Count only actions from episodes that were still active.
    for action in actions[active_before]:
        action_counts[action] += 1

    stopped_now = active_before & (actions == STOP_ACTION)
    stop_step[stopped_now] = action_eval_env.steps[stopped_now] + 1

    obs, _, _, _ = action_eval_env.step(actions)

print("Action counts across active episode steps:")
for action, count in enumerate(action_counts):
    label = "STOP" if action == STOP_ACTION else f"Toggle correction bit {action}"
    print(f"{label:24s}: {count}")

print(f"\nEpisodes stopping by STOP action: {(stop_step >= 0).sum()}/512")
print(f"Episodes reaching timeout: {(stop_step < 0).sum()}/512")

if np.any(stop_step >= 0):
    print("\nSTOP step distribution:")
    for step in sorted(np.unique(stop_step[stop_step >= 0])):
        count = int(np.sum(stop_step == step))
        print(f"Step {step}: {count} episodes")
```

```python
# Compare greedy Q-values for STOP against all toggle actions

q_device = next(shaped_agent.online_net.parameters()).device

q_env = ShapedTimedQECEnv(
    dataset=controlled_dataset,
    single_detector_effects=single_detector_effects,
    single_observable_effects=single_observable_effects,
    num_envs=512,
    max_steps=MAX_STEPS,
    stop_action=STOP_ACTION,
    seed=456,
    shaping_gamma=GAMMA,
)

q_obs = q_env.reset()
q_env.error_indices = np.arange(512)
q_env.hidden_errors = controlled_dataset["error_patterns"].copy()
q_env.hidden_logical = controlled_dataset["logical_observables"].copy()
q_env.detectors = controlled_dataset["detectors"].copy()
q_env.corrections.fill(0)
q_env.steps.fill(0)
q_env.done.fill(False)
q_env.episode_success.fill(False)
q_obs = q_env._get_observation()

def report_q_values(label, observations):
    with torch.no_grad():
        states = torch.as_tensor(
            observations, dtype=torch.float32, device=q_device
        )
        q_values = shaped_agent.online_net(states).cpu().numpy()

    best_toggle = q_values[:, :STOP_ACTION].max(axis=1)
    stop_values = q_values[:, STOP_ACTION]
    stop_greedy = stop_values >= best_toggle

    print(f"\n{label}")
    print(f"Mean Q(STOP):                 {stop_values.mean():.4f}")
    print(f"Mean best toggle Q:           {best_toggle.mean():.4f}")
    print(f"Mean (STOP - best toggle):    {(stop_values - best_toggle).mean():.4f}")
    print(f"STOP is greedy:               {int(stop_greedy.sum())}/{len(stop_greedy)}")
    print("Mean Q-value per action:")
    for action in range(NUM_ACTIONS):
        print(f"  Action {action}: {q_values[:, action].mean():.4f}")

report_q_values("Initial observations", q_obs)

# Advance one greedy action and inspect the resulting observations.
first_actions = shaped_agent.select_actions(q_obs, epsilon=0.0)
q_obs_next, _, _, _ = q_env.step(first_actions)

report_q_values("After one greedy action", q_obs_next)
```

```python
print("Unique syndrome analysis")
print("------------------------")
print("Unique syndromes:", len(unique_syndromes))
print("Active detector indices:",
      np.flatnonzero(np.any(detector_syndromes != 0, axis=0)).tolist())

print("\nSyndrome frequencies")
print("--------------------")

unique_rows, counts = np.unique(
    detector_syndromes,
    axis=0,
    return_counts=True
)

for i, (syndrome, count) in enumerate(zip(unique_rows, counts)):
    active = np.flatnonzero(syndrome)
    print(
        f"{i:2d}: detectors={active.tolist()}, "
        f"frequency={count}"
    )

# Gaussian elimination over GF(2)
def gf2_rank(matrix):
    matrix = matrix.astype(np.uint8).copy()
    rows, cols = matrix.shape
    rank = 0

    for col in range(cols):
        pivot = next(
            (r for r in range(rank, rows) if matrix[r, col]),
            None
        )

        if pivot is None:
            continue

        matrix[[rank, pivot]] = matrix[[pivot, rank]]

        for r in range(rows):
            if r != rank and matrix[r, col]:
                matrix[r] ^= matrix[rank]

        rank += 1

    return rank

print("\nGF(2) rank:", gf2_rank(unique_rows))
```

```python
# Exact algebraic decoder using GF(2)
E = shaped_train_env.single_detector_effects[:, :4].astype(np.uint8)
L = shaped_train_env.single_observable_effects.astype(np.uint8)

# Build the augmented system:
# E.T @ correction = syndrome
# correction is a 9-bit binary vector.

from itertools import product

exact_corrections = {}

for syndrome_tuple in product([0, 1], repeat=4):
    syndrome = np.array(syndrome_tuple, dtype=np.uint8)

    candidates = []

    for correction_tuple in product([0, 1], repeat=9):
        correction = np.array(correction_tuple, dtype=np.uint8)

        resulting_syndrome = (correction @ E) % 2

        if np.array_equal(resulting_syndrome, syndrome):
            logical_parity = int((correction @ L).item() % 2)

            candidates.append((correction, logical_parity))

    # Choose a deterministic correction with logical parity 0.
    valid = [
        correction
        for correction, parity in candidates
        if parity == 0
    ]

    exact_corrections[syndrome_tuple] = valid[0]

print("Exact decoder verification")
print("---------------------------")

for syndrome, correction in exact_corrections.items():
    residual = (
        np.array(syndrome, dtype=np.uint8)
        ^ ((correction @ E) % 2)
    )

    logical_parity = int((correction @ L).item() % 2)

    print(
        f"Syndrome {syndrome}: "
        f"residual={residual.tolist()}, "
        f"logical parity={logical_parity}, "
        f"correction weight={int(correction.sum())}"
    )

print("\nAll 16 syndromes have a parity-0 correction:",
      all(len(v) > 0 for v in exact_corrections.values()))
```

```python
exact_predictions = np.zeros((512, 9), dtype=np.uint8)

for i, syndrome in enumerate(detector_syndromes):
    key = tuple(syndrome[:4].astype(int))
    exact_predictions[i] = exact_corrections[key]

# Calculate residual detector syndromes.
exact_detector_effects = (
    exact_predictions @
    shaped_train_env.single_detector_effects
) % 2

exact_residual_syndromes = (
    detector_syndromes ^ exact_detector_effects
)

# Calculate residual logical observables.
exact_logical_effects = (
    exact_predictions @
    shaped_train_env.single_observable_effects
) % 2

exact_residual_logical = (
    logical_observables ^ exact_logical_effects
)

# Evaluate both success conditions.
exact_syndrome_ok = np.all(
    exact_residual_syndromes == 0, axis=1
)

exact_logical_ok = np.all(
    exact_residual_logical == 0, axis=1
)

exact_success = exact_syndrome_ok & exact_logical_ok

print("Exact decoder evaluation")
print("------------------------")
print("Total samples:", len(detector_syndromes))
print("Syndrome correction:",
      f"{exact_syndrome_ok.sum()}/{len(exact_success)}")
print("Logical correction:",
      f"{exact_logical_ok.sum()}/{len(exact_success)}")
print("Combined success:",
      f"{exact_success.sum()}/{len(exact_success)}")
print("Success rate:",
      f"{100 * exact_success.mean():.2f}%")
```

```python
print("Exact decoder correction complexity")
print("-----------------------------------")

weights = np.array([
    correction.sum()
    for correction in exact_corrections.values()
])

print("Minimum correction weight:", weights.min())
print("Maximum correction weight:", weights.max())
print("Mean correction weight:", weights.mean())
print("Median correction weight:", np.median(weights))
print("MAX_STEPS:", MAX_STEPS)

print("\nCorrection weight distribution")
print("------------------------------")

for weight in np.unique(weights):
    print(
        f"Weight {weight}: "
        f"{np.sum(weights == weight)} syndromes"
    )

print("\nSTOP feasibility:")
print("All exact corrections fit within episode:",
      bool(np.all(weights <= MAX_STEPS)))
```

```python
# Build states corresponding to exactly corrected syndromes.
# Use the exact correction vectors as diagnostic states.

diagnostic_states = []

for syndrome, correction in exact_corrections.items():
    detector = np.zeros(24, dtype=np.float32)
    detector[:4] = syndrome

    correction = correction.astype(np.float32)

    state = np.concatenate([
        detector,
        correction,
        np.array([0.1], dtype=np.float32)
    ])

    diagnostic_states.append(state)

diagnostic_states = np.asarray(
    diagnostic_states,
    dtype=np.float32
)

# Evaluate the trained online network.
agent.online_net.eval()

with torch.no_grad():
    states_tensor = torch.as_tensor(
        diagnostic_states,
        dtype=torch.float32,
        device=network_device
    )
    q_values = agent.online_net(states_tensor).cpu().numpy()

print("Q-values after exact correction")
print("--------------------------------")
print("Mean STOP Q-value:", q_values[:, 9].mean())
print("Mean best toggle Q-value:", q_values[:, :9].max(axis=1).mean())
print("States where STOP is greedy:",
      np.sum(q_values.argmax(axis=1) == 9), "/ 16")

for i, (syndrome, _) in enumerate(exact_corrections.items()):
    print(
        f"Syndrome {syndrome}: "
        f"STOP={q_values[i, 9]:.4f}, "
        f"best toggle={q_values[i, :9].max():.4f}, "
        f"greedy action={q_values[i].argmax()}"
    )
```

```python
# Select a nontrivial exact correction
correction = next(
    c for c in exact_corrections.values()
    if np.asarray(c).sum() > 0
)

correction = np.asarray(correction, dtype=np.float32)
weight = int(correction.sum())

# State after applying the correction: syndrome is now cleared
remaining_fraction = (
    shaped_train_env.max_steps - weight
) / shaped_train_env.max_steps

state = np.concatenate([
    np.zeros(24, dtype=np.float32),
    correction,
    np.array([remaining_fraction], dtype=np.float32)
]).reshape(1, -1)

device = next(shaped_agent.online_net.parameters()).device
state_tensor = torch.tensor(state, dtype=torch.float32, device=device)

shaped_agent.online_net.eval()
with torch.no_grad():
    q_values = shaped_agent.online_net(state_tensor).cpu().numpy()[0]

print("Correction weight:", weight)
print("Remaining fraction:", remaining_fraction)
print("Observation shape:", state.shape)
print("Q-values:", np.round(q_values, 4))
print("Best action:", int(np.argmax(q_values)))
print("STOP Q-value (action 9):", round(float(q_values[9]), 4))
print("Best toggle Q-value:", round(float(np.max(q_values[:9])), 4))
```
